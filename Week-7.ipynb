```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots inside the notebook
%matplotlib inline

np.random.seed(42)
df = pd.DataFrame({
    'date': pd.date_range(start='2023-01-01', periods=30),
    'temperature': np.random.normal(25, 5, 30),
    'humidity': np.random.randint(40, 90, 30),
    'wind_speed': np.random.uniform(5, 20, 30),
    'condition': np.random.choice(['Sunny', 'Rainy', 'Cloudy'], size=30)
})

# First few rows
df.head()
```

```python
# Data types
df.dtypes

# Check for missing values
df.isnull().sum()
```

```python
# Summary statistics
df.describe()

# Group by condition and compute average temperature, humidity, etc.
df.groupby('condition')[['temperature', 'humidity', 'wind_speed']].mean()
```

```python
# Visualize the data
plt.figure(figsize=(10, 5))
plt.plot(df['date'], df['temperature'], marker='o', linestyle='-')
plt.title('Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# Pairplot to visualize relationships between features
sns.pairplot(df, hue='condition', diag_kind='kde')
```

```python
#Bar plot for average temperature per weather condition
plt.figure(figsize=(8, 5))
# Calculate average temperature per condition
avg_temp = df.groupby('condition')['temperature'].mean().reset_index()

sns.barplot(data=avg_temp, x='condition', y='temperature', palette='coolwarm')
plt.title('Average Temperature per Weather Condition')
plt.xlabel('Condition')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
#Histogram of temperature distribution
plt.figure(figsize=(8, 5))
plt.hist(df['humidity'], bins=10, color='skyblue', edgecolor='black')
plt.title('Humidity Distribution')
plt.xlabel('Humidity (%)')
plt.ylabel('Frequency')
plt.show()
```

```python
#scatter plot of temperature vs humidity
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='temperature', y='humidity', hue='condition', style='condition')
plt.title('Temperature vs. Humidity')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.legend(title='Condition')
plt.show()
```

```python
# Box plot of temperature by condition
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='condition', y='temperature', palette='Set2')
plt.title('Temperature Distribution by Weather Condition')
plt.xlabel('Condition')
plt.ylabel('Temperature (°C)')
plt.show()
```
